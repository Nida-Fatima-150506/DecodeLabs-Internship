```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# TASK 2: DATA LOADING & PREPROCESSING
# =========================================================
# Load your dataset (Make sure to upload your Excel file to Colab first!)
file_name = "Dataset for Data Analytics.xlsx"
df = pd.read_excel(file_name, sheet_name="Cleaned Dataset")

print("✅ Task 2 Complete: Dataset loaded successfully!")
print(f"Dataset Shape: {df.shape[0]} rows and {df.shape[1]} columns.\n")

# =========================================================
# TASK 3: EXPLORATORY DATA ANALYSIS (EDA)
# =========================================================
print("--- TASK 3: SUMMARY STATISTICS ---")
print(f"Total Orders: {len(df)}")
print(f"Average Total Price: ${df['TotalPrice'].mean():.2f}")
print(f"Median Total Price: ${df['TotalPrice'].median():.2f}")
print(f"Maximum Total Price: ${df['TotalPrice'].max():.2f}")
print(f"Minimum Total Price: ${df['TotalPrice'].min():.2f}")

print("\n--- CATEGORICAL HIGHLIGHTS ---")
print(f"Most Popular Product: {df['Product'].mode()[0]}")
print(f"Preferred Payment Method: {df['PaymentMethod'].mode()[0]}")
print(f"Top Referral Source: {df['ReferralSource'].mode()[0]}")

# Show product counts matching your Excel pivot table
print("\n--- PRODUCT SALES COUNTS ---")
print(df['Product'].value_counts())

# =========================================================
# TASK 4: DATA VISUALIZATION
# =========================================================
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Product', order=df['Product'].value_counts().index, palette='viridis')
plt.title('Total Orders by Product Type (Week 2)', fontsize=14)
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Save the plot image
plt.savefig('product_sales_chart.png')
plt.show()
print("\n✅ Task 4 Complete: Bar chart generated and saved as 'product_sales_chart.png'!")
```
